```python
import json
from collections import Counter

KB_PATH = "/content/drive/MyDrive/setu-guide-rag/data/kb/kb_chunks.json"

with open(KB_PATH) as f:
    kb = json.load(f)

print(f"Total chunks: {len(kb)}\n")

required = {"chunk_id","stage","topic","intent","text","source"}
valid_stages = {"connect","understand","reflect","discuss","agree",
                "family","reflect_again","decision","global"}

errors = []
ids = [c["chunk_id"] for c in kb]
if len(ids) != len(set(ids)):
    errors.append("Duplicate chunk_id found!")

for c in kb:
    if required - set(c.keys()):
        errors.append(f"{c.get('chunk_id','?')} missing fields")
    if not c.get("text","").strip():
        errors.append(f"{c['chunk_id']} empty text")
    if c["stage"] not in valid_stages:
        errors.append(f"{c['chunk_id']} invalid stage: {c['stage']}")

print("Chunks per stage:")
for s, n in sorted(Counter(c["stage"] for c in kb).items()):
    print(f"  {s:15s}: {n}")

print("\nChunks per intent:")
for i, n in sorted(Counter(c["intent"] for c in kb).items()):
    print(f"  {i:22s}: {n}")

if errors:
    print("\n❌ ERRORS:")
    for e in errors: print("  -", e)
else:
    print("\n✅ All checks passed.")
```
